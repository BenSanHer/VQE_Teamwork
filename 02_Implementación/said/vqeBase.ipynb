```python
from defs import *  
import pennylane as qml
from pennylane import numpy as np
import matplotlib.pyplot as plt
from qiskit_ibm_runtime import QiskitRuntimeService
```

#### A) Define the Hamiltonian

```python
'''H = Z_0*Z1 + X_0*X1, cuyo estado fundamental (energía mínima) 
conozco teóricamente que es -2.0.
Este H es el modelo de Heisenberg para dos qubits
# @ == producto tensorial'''
obs = [qml.PauliZ(0) @ qml.PauliZ(1), qml.PauliX(0) @ qml.PauliX(1)]
coefs = [1.0, 1.0] # Para captar la fuerza de interacción entre los términos, que en moléculas tiene qu ever con la distancia de los electrones.
# Queda H=1.0(Z_0​⊗Z_1​ )+1.0(X_0​⊗X_1​)
H = qml.Hamiltonian(coefs, obs)
```

#### B) Define angles

```python
# Esta parte es la que creo que se puede hacer con IA
angulos = np.array([-1.0, 2.5])
```

#### C) Ansatzs

```python
# Experimento 1: El Ansatz bueno
print(f"EXPERIMENTO 1: Ansatz")
params_buenos = ejecutar_vqe_local(Hamiltonian=H, angles=angulos, opcion_ansatz=1, pasos=50)
```
